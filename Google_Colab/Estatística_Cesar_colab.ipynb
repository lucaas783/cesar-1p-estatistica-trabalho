```
# ==========================================================
# DISTRIBUIÇÃO DOS RESPONDENTES POR INSTITUIÇÃO
# ==========================================================

import pandas as pd
import plotly.express as px

# ==========================================================
# CORES INSTITUCIONAIS
# ==========================================================

cores = {
    "UFPE": "#0057B8",
    "CESAR School": "#F58220",
    "UPE": "#009739",
    "Outras Instituições": "#808080"
}

# ==========================================================
# DADOS
# ==========================================================

dados = pd.DataFrame({
    "Instituição de Ensino": [
        "UFPE",
        "CESAR School",
        "UPE",
        "Outras Instituições"
    ],
    "Número de Respondentes": [
        118,
        38,
        16,
        11
    ]
})

# ==========================================================
# CÁLCULO DOS PERCENTUAIS
# ==========================================================

total_respondentes = dados["Número de Respondentes"].sum()

dados["Percentual"] = (
    dados["Número de Respondentes"]
    / total_respondentes
    * 100
)

# ==========================================================
# GRÁFICO
# ==========================================================

fig = px.bar(
    dados,
    x="Instituição de Ensino",
    y="Número de Respondentes",
    color="Instituição de Ensino",
    color_discrete_map=cores,
    text=dados["Percentual"].round(1),
    title="🎓 Distribuição dos Respondentes por Instituição de Ensino"
)

fig.update_traces(
    texttemplate="%{text}%",
    textposition="outside"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    height=600,
    showlegend=False,
    xaxis_title="",
    yaxis_title="Número de Respondentes",
    font=dict(
        family="Arial",
        size=14
    )
)

fig.show()
```

```
# ==========================================================
# DASHBOARD GERAL DOS RESPONDENTES
# ==========================================================

import pandas as pd
from plotly.subplots import make_subplots
import plotly.graph_objects as go

# ==========================================================
# DADOS
# ==========================================================

faculdades = pd.DataFrame({
    "Categoria": ["UFPE", "CESAR School", "UPE", "Outras Instituições"],
    "Quantidade": [118, 38, 16, 11]
})

esportes = pd.DataFrame({
    "Categoria": [
        "Futebol",
        "Outras modalidades",
        "Vôlei",
        "Automobilismo",
        "Nenhum",
        "Basquete"
    ],
    "Quantidade": [99, 25, 22, 15, 14, 8]
})

transportes = pd.DataFrame({
    "Categoria": [
        "Transporte Público",
        "Veículo Particular",
        "Aplicativos de Mobilidade",
        "Deslocamento a Pé",
        "Outras Modalidades"
    ],
    "Quantidade": [113, 46, 11, 7, 6]
})

# ==========================================================
# PERCENTUAIS
# ==========================================================

faculdades["Percentual"] = (
    faculdades["Quantidade"]
    / faculdades["Quantidade"].sum()
    * 100
)

esportes["Percentual"] = (
    esportes["Quantidade"]
    / esportes["Quantidade"].sum()
    * 100
)

transportes["Percentual"] = (
    transportes["Quantidade"]
    / transportes["Quantidade"].sum()
    * 100
)

# ==========================================================
# DASHBOARD
# ==========================================================

fig = make_subplots(
    rows=1,
    cols=3,
    subplot_titles=(
        "🎓 Instituições de Ensino",
        "⚽ Preferências Esportivas",
        "🚌 Modalidades de Transporte"
    )
)

# ----------------------------------------------------------
# FACULDADES
# ----------------------------------------------------------

fig.add_trace(
    go.Bar(
        x=faculdades["Categoria"],
        y=faculdades["Quantidade"],
        text=[
            f"{p:.1f}%"
            for p in faculdades["Percentual"]
        ],
        textposition="outside",
        marker_color=[
            "#0057B8",
            "#F58220",
            "#009739",
            "#808080"
        ],
        showlegend=False
    ),
    row=1,
    col=1
)

# ----------------------------------------------------------
# ESPORTES
# ----------------------------------------------------------

fig.add_trace(
    go.Bar(
        x=esportes["Categoria"],
        y=esportes["Quantidade"],
        text=[
            f"{p:.1f}%"
            for p in esportes["Percentual"]
        ],
        textposition="outside",
        marker_color="#0F4C5C",
        showlegend=False
    ),
    row=1,
    col=2
)

# ----------------------------------------------------------
# TRANSPORTE
# ----------------------------------------------------------

fig.add_trace(
    go.Bar(
        x=transportes["Categoria"],
        y=transportes["Quantidade"],
        text=[
            f"{p:.1f}%"
            for p in transportes["Percentual"]
        ],
        textposition="outside",
        marker_color="#2E8B57",
        showlegend=False
    ),
    row=1,
    col=3
)

# ==========================================================
# LAYOUT
# ==========================================================

fig.update_layout(
    template="plotly_white",
    title={
        "text": "📊 Panorama Geral dos Respondentes",
        "x": 0.5
    },
    height=650,
    width=1500,
    font=dict(
        family="Arial",
        size=13
    ),
    margin=dict(
        t=100,
        l=40,
        r=40,
        b=40
    )
)

fig.update_yaxes(
    title_text="Número de Respondentes"
)

fig.show()
```

```
# ==========================================================
# HEATMAP - PERFIL DE TRANSPORTE POR INSTITUIÇÃO
# ==========================================================

import pandas as pd
import plotly.express as px

# ==========================================================
# DADOS CORRETOS DE TRANSPORTE
# ==========================================================

dados_transporte = [
    ("CESAR", "Carro", 16),
    ("CESAR", "Transporte Público", 13),
    ("CESAR", "APP", 6),
    ("CESAR", "Outros", 3),

    ("UFPE", "Transporte Público", 79),
    ("UFPE", "Carro", 25),
    ("UFPE", "A pé", 7),
    ("UFPE", "APP", 4),
    ("UFPE", "Outros", 3),

    ("UPE", "Transporte Público", 13),
    ("UPE", "Carro", 3)
]

df_plot = pd.DataFrame(
    dados_transporte,
    columns=[
        "Faculdade",
        "Transporte",
        "Quantidade"
    ]
)

# ==========================================================
# TABELA DINÂMICA
# ==========================================================

pivot = df_plot.pivot_table(
    index="Faculdade",
    columns="Transporte",
    values="Quantidade",
    fill_value=0
)

# Ordenação opcional das colunas
ordem_transportes = [
    "Transporte Público",
    "Carro",
    "APP",
    "A pé",
    "Outros"
]

pivot = pivot.reindex(
    columns=ordem_transportes,
    fill_value=0
)

# ==========================================================
# HEATMAP
# ==========================================================

fig = px.imshow(
    pivot,
    text_auto=True,
    color_continuous_scale="Blues",
    aspect="auto",
    title="🚌 Perfil das Modalidades de Transporte por Instituição de Ensino"
)

fig.update_layout(
    template="plotly_white",
    title={
        "text": "🚌 Perfil das Modalidades de Transporte por Instituição de Ensino",
        "x": 0.5
    },
    height=600,
    width=1000,
    font=dict(
        family="Arial",
        size=13
    ),
    coloraxis_colorbar=dict(
        title="Respondentes"
    )
)

fig.update_xaxes(
    title="Modalidade de Transporte"
)

fig.update_yaxes(
    title="Instituição de Ensino"
)

fig.show()

# ==========================================================
# TABELA UTILIZADA NO HEATMAP
# ==========================================================

print("\nPERFIL DE TRANSPORTE POR FACULDADE")
print("="*60)
print(pivot)
```

```
# ==========================================================
# VISÃO GERAL DOS RESPONDENTES
# ==========================================================

import pandas as pd
from plotly.subplots import make_subplots
import plotly.graph_objects as go

# ==========================================================
# DADOS
# ==========================================================

faculdades = pd.DataFrame({
    "Categoria": [
        "UFPE",
        "CESAR",
        "UPE",
        "Outras Instituições"
    ],
    "Quantidade": [
        118,
        38,
        16,
        11
    ]
})

esportes = pd.DataFrame({
    "Categoria": [
        "Futebol",
        "Outras Modalidades",
        "Voleibol",
        "Automobilismo",
        "Sem Preferência",
        "Basquetebol"
    ],
    "Quantidade": [
        99,
        25,
        22,
        15,
        14,
        8
    ]
})

transportes = pd.DataFrame({
    "Categoria": [
        "Transporte Público",
        "Veículo Particular",
        "Aplicativos de Mobilidade",
        "Deslocamento a Pé",
        "Outras Modalidades"
    ],
    "Quantidade": [
        113,
        46,
        11,
        7,
        6
    ]
})

# ==========================================================
# PERCENTUAIS
# ==========================================================

for df in [faculdades, esportes, transportes]:
    df["Percentual"] = (
        df["Quantidade"] /
        df["Quantidade"].sum()
    ) * 100

# ==========================================================
# DASHBOARD
# ==========================================================

fig = make_subplots(
    rows=1,
    cols=3,
    subplot_titles=(
        "🎓 Instituições de Ensino",
        "⚽ Preferências Esportivas",
        "🚌 Modalidades de Transporte"
    ),
    horizontal_spacing=0.12
)

# ==========================================================
# FACULDADES
# ==========================================================

fig.add_trace(
    go.Bar(
        y=faculdades["Categoria"],
        x=faculdades["Percentual"],
        orientation="h",
        text=[
            f"{p:.1f}% ({q})"
            for p, q in zip(
                faculdades["Percentual"],
                faculdades["Quantidade"]
            )
        ],
        textposition="outside",
        marker_color="#4C78A8",
        showlegend=False
    ),
    row=1,
    col=1
)

# ==========================================================
# ESPORTES
# ==========================================================

fig.add_trace(
    go.Bar(
        y=esportes["Categoria"],
        x=esportes["Percentual"],
        orientation="h",
        text=[
            f"{p:.1f}% ({q})"
            for p, q in zip(
                esportes["Percentual"],
                esportes["Quantidade"]
            )
        ],
        textposition="outside",
        marker_color="#F58220",
        showlegend=False
    ),
    row=1,
    col=2
)

# ==========================================================
# TRANSPORTE
# ==========================================================

fig.add_trace(
    go.Bar(
        y=transportes["Categoria"],
        x=transportes["Percentual"],
        orientation="h",
        text=[
            f"{p:.1f}% ({q})"
            for p, q in zip(
                transportes["Percentual"],
                transportes["Quantidade"]
            )
        ],
        textposition="outside",
        marker_color="#009739",
        showlegend=False
    ),
    row=1,
    col=3
)

# ==========================================================
# LAYOUT
# ==========================================================

fig.update_layout(
    template="plotly_white",
    title={
        "text": "📊 Visão Geral dos Respondentes",
        "x": 0.5
    },
    height=700,
    width=1600,
    font=dict(
        family="Arial",
        size=13
    ),
    margin=dict(
        t=90,
        l=30,
        r=30,
        b=30
    )
)

fig.update_xaxes(
    title_text="Percentual (%)"
)

fig.show()

# ==========================================================
# RESUMO ESTATÍSTICO
# ==========================================================

print("\n" + "=" * 70)
print("RESUMO GERAL DA PESQUISA")
print("=" * 70)

print("\n🎓 DISTRIBUIÇÃO DAS INSTITUIÇÕES")
print(faculdades[["Categoria", "Quantidade"]])

print("\n⚽ PREFERÊNCIAS ESPORTIVAS")
print(esportes[["Categoria", "Quantidade"]])

print("\n🚌 MODALIDADES DE TRANSPORTE")
print(transportes[["Categoria", "Quantidade"]])
```

```
# ==========================================================
# HEATMAP - PREFERÊNCIAS ESPORTIVAS POR INSTITUIÇÃO
# ==========================================================

import pandas as pd
import plotly.express as px

# ==========================================================
# DADOS
# ==========================================================

dados_esportes = [

    # CESAR
    ("CESAR", "Futebol", 22),
    ("CESAR", "Outras Modalidades", 9),
    ("CESAR", "Voleibol", 3),
    ("CESAR", "Automobilismo", 2),
    ("CESAR", "Basquetebol", 1),
    ("CESAR", "Sem Preferência", 1),

    # UFPE
    ("UFPE", "Futebol", 61),
    ("UFPE", "Voleibol", 15),
    ("UFPE", "Outras Modalidades", 13),
    ("UFPE", "Automobilismo", 12),
    ("UFPE", "Sem Preferência", 11),
    ("UFPE", "Basquetebol", 6),

    # UPE
    ("UPE", "Futebol", 9),
    ("UPE", "Voleibol", 3),
    ("UPE", "Automobilismo", 1),
    ("UPE", "Basquetebol", 1),
    ("UPE", "Outras Modalidades", 1),
    ("UPE", "Sem Preferência", 1)
]

df = pd.DataFrame(
    dados_esportes,
    columns=[
        "Instituição",
        "Modalidade Esportiva",
        "Quantidade"
    ]
)

# ==========================================================
# TABELA PARA HEATMAP
# ==========================================================

pivot = df.pivot_table(
    index="Instituição",
    columns="Modalidade Esportiva",
    values="Quantidade",
    fill_value=0
)

ordem_esportes = [
    "Futebol",
    "Voleibol",
    "Automobilismo",
    "Basquetebol",
    "Sem Preferência",
    "Outras Modalidades"
]

pivot = pivot.reindex(
    columns=ordem_esportes
)

# ==========================================================
# HEATMAP
# ==========================================================

fig = px.imshow(
    pivot,
    text_auto=True,
    color_continuous_scale="YlOrRd",
    aspect="auto",
    title="⚽ Mapa de Calor das Preferências Esportivas por Instituição"
)

fig.update_layout(
    template="plotly_white",
    title={
        "text": "⚽ Mapa de Calor das Preferências Esportivas por Instituição",
        "x": 0.5
    },
    height=550,
    width=1100,
    font=dict(
        family="Arial",
        size=14
    ),
    coloraxis_colorbar=dict(
        title="Respondentes"
    )
)

fig.update_xaxes(
    title_text="Modalidade Esportiva"
)

fig.update_yaxes(
    title_text="Instituição"
)

fig.show()

# ==========================================================
# TABELA UTILIZADA
# ==========================================================

print("\n" + "=" * 70)
print("PERFIL ESPORTIVO POR INSTITUIÇÃO")
print("=" * 70)

print(pivot)
```

```
# ==========================================================
# ANÁLISE VISUAL DOS RESPONDENTES
# Esportes + Transporte por Faculdade
# ==========================================================

import pandas as pd
import plotly.express as px

# ==========================================================
# CORES INSTITUCIONAIS
# ==========================================================

cores = {
    "CESAR School": "#F58220",  # Laranja
    "UFPE": "#0057B8",          # Azul
    "UPE": "#009739"            # Verde
}

# ==========================================================
# DADOS
# ==========================================================

dados = [

    # -------------------------
    # ESPORTES - CESAR
    # -------------------------

    ("CESAR School", "Esporte", "Futebol", 22),
    ("CESAR School", "Esporte", "Artes Marciais", 5),
    ("CESAR School", "Esporte", "Vôlei", 3),
    ("CESAR School", "Esporte", "Automobilismo", 2),
    ("CESAR School", "Esporte", "Outros", 6),

    # -------------------------
    # ESPORTES - UFPE
    # -------------------------

    ("UFPE", "Esporte", "Futebol", 61),
    ("UFPE", "Esporte", "Vôlei", 15),
    ("UFPE", "Esporte", "Automobilismo", 12),
    ("UFPE", "Esporte", "Nenhum", 11),
    ("UFPE", "Esporte", "Outros", 19),

    # -------------------------
    # ESPORTES - UPE
    # -------------------------

    ("UPE", "Esporte", "Futebol", 9),
    ("UPE", "Esporte", "Vôlei", 3),
    ("UPE", "Esporte", "Automobilismo", 1),
    ("UPE", "Esporte", "Basquete", 1),
    ("UPE", "Esporte", "Outros", 2),

    # -------------------------
    # TRANSPORTE - CESAR
    # -------------------------

    ("CESAR School", "Transporte", "Carro", 16),
    ("CESAR School", "Transporte", "Transporte Público", 13),
    ("CESAR School", "Transporte", "APP", 6),
    ("CESAR School", "Transporte", "Moto", 2),
    ("CESAR School", "Transporte", "Outros", 1),

    # -------------------------
    # TRANSPORTE - UFPE
    # -------------------------

    ("UFPE", "Transporte", "Transporte Público", 72),
    ("UFPE", "Transporte", "Carro", 31),
    ("UFPE", "Transporte", "APP", 8),
    ("UFPE", "Transporte", "A Pé", 4),
    ("UFPE", "Transporte", "Outros", 3),

    # -------------------------
    # TRANSPORTE - UPE
    # -------------------------

    ("UPE", "Transporte", "Transporte Público", 10),
    ("UPE", "Transporte", "Carro", 5),
    ("UPE", "Transporte", "Moto", 1),
]

# ==========================================================
# DATAFRAME
# ==========================================================

df = pd.DataFrame(
    dados,
    columns=[
        "Faculdade",
        "Categoria",
        "Item",
        "Quantidade"
    ]
)

# ==========================================================
# ESPORTES (%)
# ==========================================================

df_esporte = df[df["Categoria"] == "Esporte"].copy()

df_esporte["Percentual"] = (
    df_esporte.groupby("Faculdade")["Quantidade"]
    .transform(lambda x: x / x.sum() * 100)
)

fig1 = px.bar(
    df_esporte,
    x="Item",
    y="Percentual",
    color="Faculdade",
    color_discrete_map=cores,
    barmode="group",
    text=df_esporte["Percentual"].round(1),
    title="⚽ Preferências Esportivas por Faculdade (%)"
)

fig1.update_traces(
    texttemplate="%{text}%",
    textposition="outside"
)

fig1.update_layout(
    template="plotly_white",
    title_x=0.5,
    height=650,
    legend_title="Faculdade",
    xaxis_title="",
    yaxis_title="Percentual (%)"
)

fig1.show()

# ==========================================================
# TRANSPORTE (%)
# ==========================================================

df_transporte = df[df["Categoria"] == "Transporte"].copy()

df_transporte["Percentual"] = (
    df_transporte.groupby("Faculdade")["Quantidade"]
    .transform(lambda x: x / x.sum() * 100)
)

fig2 = px.bar(
    df_transporte,
    x="Item",
    y="Percentual",
    color="Faculdade",
    color_discrete_map=cores,
    barmode="group",
    text=df_transporte["Percentual"].round(1),
    title="🚗 Meio de Transporte por Faculdade (%)"
)

fig2.update_traces(
    texttemplate="%{text}%",
    textposition="outside"
)

fig2.update_layout(
    template="plotly_white",
    title_x=0.5,
    height=650,
    legend_title="Faculdade",
    xaxis_title="",
    yaxis_title="Percentual (%)"
)

fig2.show()

# ==========================================================
# TREEMAP
# ==========================================================

fig3 = px.treemap(
    df,
    path=[
        "Faculdade",
        "Categoria",
        "Item"
    ],
    values="Quantidade",
    color="Faculdade",
    color_discrete_map=cores,
    title="📊 Perfil Geral dos Respondentes"
)

fig3.update_layout(
    template="plotly_white",
    title_x=0.5,
    height=700
)

fig3.show()

# ==========================================================
# SUNBURST
# ==========================================================

fig4 = px.sunburst(
    df,
    path=[
        "Faculdade",
        "Categoria",
        "Item"
    ],
    values="Quantidade",
    color="Faculdade",
    color_discrete_map=cores,
    title="🌎 Visão Geral dos Respondentes"
)

fig4.update_layout(
    template="plotly_white",
    title_x=0.5,
    height=750
)

fig4.show()

# ==========================================================
# INSIGHTS AUTOMÁTICOS
# ==========================================================

esporte_mais_comum = (
    df[df["Categoria"] == "Esporte"]
    .groupby("Item")["Quantidade"]
    .sum()
    .idxmax()
)

transporte_mais_comum = (
    df[df["Categoria"] == "Transporte"]
    .groupby("Item")["Quantidade"]
    .sum()
    .idxmax()
)

print("\n" + "=" * 60)
print("INSIGHTS DA PESQUISA")
print("=" * 60)

print(f"\n🏆 Esporte mais popular: {esporte_mais_comum}")
print(f"🚍 Transporte mais utilizado: {transporte_mais_comum}")

print("\nResumo:")
print(
    "O futebol apresenta ampla predominância entre os estudantes, "
    "enquanto o transporte público se destaca como principal meio "
    "de deslocamento. A UFPE concentra a maior parte dos respondentes, "
    "mas as tendências observadas permanecem semelhantes entre as instituições."
)
```
